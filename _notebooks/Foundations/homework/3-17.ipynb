## Popcorn

### Popcorn Checkpoint: Expand the Catalog

**Prediction:** For 8 books, I predict that linear search will require 8 checks in the worst case, while binary search will require fewer checks because it repeatedly cuts the search area in half.

````python
# CODE_RUNNER: Library 3.17 Popcorn - Expand the catalog

import math

n = 8

linear_checks = n
binary_checks = 0 if n == 0 else math.floor(math.log2(n)) + 1

print(n)
print(linear_checks)
print(binary_checks)
````

**Actual output:**

8
8
4

After changing `n` to 16:

````python
# CODE_RUNNER: Library 3.17 Popcorn - Expanded catalog

import math

n = 16

linear_checks = n
binary_checks = 0 if n == 0 else math.floor(math.log2(n)) + 1

print(n)
print(linear_checks)
print(binary_checks)
````

**Actual output:**

16
16
5

**Explanation:** When the number of books doubles from 8 to 16, the linear-search worst-case checks double from 8 to 16. The binary-search checks only increase from 4 to 5 because binary search repeatedly halves the number of possible books.

**Recommendation:** For a large catalog that is already sorted and allows direct access to the middle entry, binary search can find a book with fewer worst-case checks. However, binary search requires the catalog to be sorted.

### 4. Recommend a method for repeated searches

For 1,000 books, the supplied model gives:

- Binary search worst case: **10 checks**

For repeated searches in an already sorted catalog, binary search can require far fewer checks in the worst case. However, binary search depends on the catalog being sorted and having direct access to middle entries.

If the catalog is unsorted, it would need to be sorted first before binary search could be used. Sorting adds additional work. If the librarian stores an extra sorted copy of the catalog, that also uses additional memory.

The important tradeoff is that binary search can reduce search time for a sorted catalog, while maintaining another sorted copy can require additional memory.

## Design Thinking

### Librarian Need

A librarian needs to find books efficiently as the catalog grows.

### Problem

Different correct algorithms can require very different amounts of work. A method that works well for a small catalog may not scale as well when the number of books becomes much larger.

### Methods Considered

- **Linear search:** Checks entries one at a time.
- **Binary search:** Repeatedly halves a sorted catalog.
- **Quadratic comparison:** Compares every book with every book when the task requires pairs.

### Prototype

I created simple calculators that use the number of books `n` to model the amount of work required.

### Test

I tested different values of `n`, including 12, 24, and 0. I also compared the supplied search estimates for a catalog of 1,000 books.

### Revision

The results showed that doubling `n` doubles linear work but quadruples quadratic work. Binary search grows much more slowly than linear search, but it requires a sorted catalog and direct access to entries.

The final recommendation is based on the specific task and requirements rather than simply choosing the algorithm with the shortest code.
