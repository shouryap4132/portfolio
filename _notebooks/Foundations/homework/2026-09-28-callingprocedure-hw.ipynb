## Popcorn Hack 2: Decomposition & Helper Procedures

**Task:** Define a helper procedure called `is_priority(incident_type)` that returns `True` for a medical emergency. Then define `process_report(location, incident_type)` that calls the helper procedure and prints a different message depending on whether the incident is a priority.

```python
# CODE_RUNNER: Popcorn Hack 2 - Decomposition & Helper Procedures

# Helper Procedure
def is_priority(incident_type):
    return incident_type == "Medical Emergency"

# Outer Procedure
def process_report(location, incident_type):
    priority = is_priority(incident_type)

    if priority:
        print("PRIORITY: " + incident_type + " reported in " + location)
    else:
        print("STANDARD REPORT: " + incident_type + " reported in " + location)

# Call the outer procedure twice with different incident types
process_report("Poway", "Medical Emergency")
process_report("Poway", "Power Outage")
```

## Homework Hack: Emergency Response Dispatch

**Task:** Create a simplified emergency-response program.

The program must:

1. Define a helper procedure called `check_priority(incident_type)`.
2. Define an outer procedure called `dispatch_response(location, incident_type)`.
3. Have `dispatch_response()` call `check_priority()`.
4. Have `check_priority()` return a Boolean value.
5. Have `dispatch_response()` print a different response depending on the returned value.
6. Call `dispatch_response()` at least three times with different incident types.
7. Use at least two different locations or incident types.

```python
# CODE_RUNNER: Homework - Emergency Response Dispatch

# Helper Procedure
def check_priority(incident_type):
    return incident_type == "Medical Emergency"

# Outer Procedure
def dispatch_response(location, incident_type):
    priority = check_priority(incident_type)

    if priority:
        print("PRIORITY RESPONSE: " + incident_type + " reported in " + location)
    else:
        print("STANDARD RESPONSE: " + incident_type + " reported in " + location)

# Test the procedure with different locations and incident types
dispatch_response("Poway", "Medical Emergency")
dispatch_response("San Diego", "Power Outage")
dispatch_response("Rancho Bernardo", "Medical Emergency")
```
